```python
# Gradient Descent to minimize f(x) = 5x^4 + 3x^2 + 10
def gradient_descent(learning_rate, max_iterations, tolerance):
    # Define the derivative of f(x): f'(x) = 15x^2 + 6x
    def derivative(x):
        return 15 * x ** 2 + 6 * x

    # Initialize x to a starting value (e.g., x = 0)
    x = 0

    for i in range(max_iterations):
        # Calculate the gradient (slope)
        grad = derivative(x)

        # Update x using the learning rate
        new_x = x - learning_rate * grad

        # Check for convergence
        if abs(new_x - x) < tolerance:
            break

        # Update x for the next iteration
        x = new_x

    return x

# Parameters
learning_rate = 0.01
max_iterations = 10000
tolerance = 1e-6

# Find the value of x that minimizes f(x)
x_min = gradient_descent(learning_rate, max_iterations, tolerance)
print("The value of x that minimizes f(x):", x_min)
```

```python
# Gradient Descent to minimize g(x, y) = 3x^2 + 5e^(-y) + 10
def gradient_descent_2d(learning_rate, max_iterations, tolerance):
    # Define the partial derivatives of g(x, y):
    # dg/dx = 6x, dg/dy = -5e^(-y)
    def partial_derivatives(x, y):
        dg_dx = 6 * x
        dg_dy = -5 * (2.718281828459045 ** -y)  # Approximation of e^(-y)
        return dg_dx, dg_dy

    # Initialize x and y to starting values (e.g., x = 0, y = 0)
    x, y = 0, 0

    for i in range(max_iterations):
        # Calculate the gradients
        grad_x, grad_y = partial_derivatives(x, y)

        # Update x and y using the learning rate
        new_x = x - learning_rate * grad_x
        new_y = y - learning_rate * grad_y

        # Check for convergence
        if abs(new_x - x) < tolerance and abs(new_y - y) < tolerance:
            break

        # Update x and y for the next iteration
        x, y = new_x, new_y

    return x, y

# Parameters
learning_rate = 0.01
max_iterations = 10000
tolerance = 1e-6

# Find the values of x and y that minimize g(x, y)
x_min, y_min = gradient_descent_2d(learning_rate, max_iterations, tolerance)
print("The values of x and y that minimize g(x, y):", x_min, y_min)
```

```python
# Gradient Descent to minimize z(x) = 1 / (1 + e^(-x))
def gradient_descent_sigmoid(learning_rate, max_iterations, tolerance):
    # Define the derivative of z(x): z'(x) = z(x) * (1 - z(x))
    def sigmoid_derivative(x):
        sigmoid = 1 / (1 + 2.718281828459045 ** -x)  # Approximation of e^(-x)
        return sigmoid * (1 - sigmoid)

    # Initialize x to a starting value (e.g., x = 0)
    x = 0

    for i in range(max_iterations):
        # Calculate the gradient (slope)
        grad = sigmoid_derivative(x)

        # Update x using the learning rate
        new_x = x - learning_rate * grad

        # Check for convergence
        if abs(new_x - x) < tolerance:
            break

        # Update x for the next iteration
        x = new_x

    return x

# Parameters
learning_rate = 0.01
max_iterations = 10000
tolerance = 1e-6

# Find the value of x that minimizes z(x)
x_min = gradient_descent_sigmoid(learning_rate, max_iterations, tolerance)
print("The value of x that minimizes z(x):", x_min)
```
